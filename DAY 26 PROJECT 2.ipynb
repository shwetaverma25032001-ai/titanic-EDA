# 🚢 Titanic Dataset - Exploratory Data Analysis (EDA)

### Data Analysis using Python, NumPy and Pandas
## 📚 Step 1: Import Libraries

#### Import the required Python libraries such as Pandas and NumPy for data analysis and data cleaning.

```python
import pandas as pd
import numpy as np
```

## 📂 Step 2: Load the Dataset

### Load the Titanic dataset into a Pandas DataFrame for analysis.

```python
df = pd.read_csv("Titanic.csv")
```

## 🔍 Step 3: Explore the Dataset

#### Display the first few rows of the dataset using `head()` and understand the structure of the dataset using `info()`.

```python
print("===== First 5 Rows of my data =====")
print(df.head())

# 3. Dataset Information
print("\n===== INFO OF MY DATA =====")
print(df.info())
```

## 🧹 Step 4: Check and Clean Missing Values

### Identify missing values using `isnull()` and `isnull().sum()`. Remove the `Cabin` column because it contains many missing values. Fill missing values in the `Age` column using the median.

```python
# 4. Check Null Values
print("\n===== NULL VALUES IN THE DATA =====")
print(df.isnull())

# 5. Count Null Values
print("\n===== NULL VALUE COUNT IN THE DATA =====")
print(df.isnull().sum())
# 4. Check Null Values
print("\n===== NULL VALUES IN THE DATA =====")
print(df.isnull())

# 5. Count Null Values
print("\n===== NULL VALUE COUNT IN THE DATA =====")
print(df.isnull().sum())

# Data Cleaning

# 6. Drop Cabin Column
df.drop("Cabin", axis=1, inplace=True)

print("\n===== COLUMNS AFTER DROPPING CABIN =====")
print(df.columns)

# 7. Find Median of Age
age_median = df["Age"].median()
print("\nMedian Age =", age_median)

# 8. Fill Missing Age with Median
df["Age"] = df["Age"].fillna(age_median)

# 9. Check Again
print("\n===== NULL VALUES AFTER FILLING AGE =====")
print(df.isnull().sum())
```

## Step 5: Data Filtering

### Filter passengers based on different conditions, such as children (Age < 18), male passengers, and female passengers.

```python
# 10. Children (Age < 18)
children = df[df["Age"] < 18]

print("\n===== CHILDREN (AGE < 18) =====")
print(children)

print("Total Children :", children.shape[0])

# 11. Female Passengers
female = df[df["Sex"] == "female"]

print("\n===== FEMALE PASSENGERS =====")
print(female)

print("Total Female :", female.shape[0])

# 12. Male Passengers
male = df[df["Sex"] == "male"]

print("\n===== MALE PASSENGERS =====")
print(male)

print("Total Male :", male.shape[0])

# 13. Survival Percentage
survival_rate = df["Survived"].mean() * 100

print("\n===== SURVIVAL RATE =====")
print("Survival Rate = {:.2f}%".format(survival_rate))
```

## 📊 Step 6: GroupBy and Aggregation

### Use GroupBy and aggregation functions to analyze the average age, average fare, and survival rate for different passenger classes and genders.

```python
# 14. Survival by Gender
print("\n===== SURVIVAL BY GENDER =====")
print(df.groupby("Sex")["Survived"].mean())

# 15. Average Age by Passenger Class
print("\n===== AVERAGE AGE BY CLASS =====")
print(df.groupby("Pclass")["Age"].mean())

# 16. Multiple Aggregation
print("\n===== GROUPBY + AGGREGATION =====")
print(
    df.groupby("Pclass").agg({
        "Age": "mean",
        "Fare": "mean",
        "Survived": "mean"
    })
)
```

## 💵 Step 7: Fare Analysis

### Find the passenger who paid the highest fare and compare the average fare across different passenger classes.

```python
# Fare Analysis
# 17. Passenger Who Paid Highest Fare
print("\n===== PASSENGER WHO PAID HIGHEST FARE =====")
print(df.loc[df["Fare"].idxmax()])

# 18. Average Fare by Passenger Class
print("\n===== AVERAGE FARE BY CLASS =====")
print(df.groupby("Pclass")["Fare"].mean())

# Final Shape
print("\n===== FINAL DATASET SHAPE =====")
print(df.shape)

print("\n===== EDA COMPLETED SUCCESSFULLY =====")

```
